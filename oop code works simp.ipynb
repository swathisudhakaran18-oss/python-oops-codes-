```python
#oops language 
# Base class (Parent)
class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def introduce(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."


# Derived class (Child class)
class Student(Person):
    def __init__(self, name: str, age: int, major: str):
        # Call the parent constructor to initialize name and age
        super().__init__(name, age)
        # Initialize child-specific attributes
        self.major = major

    # Specific behavior for Student
    def study(self):
        return f"{self.name} is studying {self.major}."


# Derived class (Child class)
class Teacher(Person):
    def __init__(self, name: str, age: int,subject: str):
        # Call the parent constructor
        super().__init__(name, age)
        # Initialize child-specific attributes
        self.subject = subject

    # Specific behavior for Teacher
    def teach(self):
        return f"Professor {self.name} is teaching {self.subject}."
        
# Creating instances (objects)
student = Student(name="swathi", age=25, major="Computer Science")
teacher = Teacher(name="ashwin", age=30, subject="Mathematics")

# Accessing inherited methods
print(student.introduce())  
print(teacher.introduce())  

# Accessing class-specific methods
print(student.study())     
print(teacher.teach())      
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner       # Instance variable for account holder's name
        self.balance = balance   # Instance variable for account balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited ₹{amount}. New balance: ₹{self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ₹{amount}. Remaining balance: ₹{self.balance}")
        else:
            print("Insufficient balance!")

# --- Usage ---
acc1 = BankAccount("Alice", 5000)
acc2 = BankAccount("Bob", 2000)

acc1.deposit(1500)
acc2.withdraw(1000)
```

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. Remaining balance: {self.balance}")
        else:
            print("Invalid withdrawal amount.")

# Derived class (inherits from Account)
class SavingsAccount(Account):
    def __init__(self, owner, balance, interest_rate):
        # Initialize parent attributes using super()
        super().__init__(owner, balance)
        self.interest_rate = interest_rate
    def add_interest(self):
        interest = self.balance * self.interest_rate / 100
        self.balance += interest
        print(f"Interest of {interest} added. New balance: {self.balance}")


acc = SavingsAccount("swathi", 1000, 5)
acc.deposit(500)
acc.add_interest() 
acc.withdraw(200)
```
